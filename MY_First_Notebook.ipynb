```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1,100,10_00_000)
y = np.sqrt(x)
plt.plot(x,y , label = "y = squareroot of x")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid 
plt.show()
```
