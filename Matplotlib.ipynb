```python


import matplotlib.pyplot as plt


oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]


import numpy as np

x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2,oscar_revenue,width,label = "oscar_movie")
plt.bar(x+width/2,non_oscar_revenue,width,label = "non_oscar_movie")
plt.title("oscar vs non oscar")
plt.xlabel("years")
plt.ylabel("revenue")
plt.legend()
plt.tight_layout()
plt.xticks(x,years)
```
